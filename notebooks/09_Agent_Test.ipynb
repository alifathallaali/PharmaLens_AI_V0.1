```python
def business_summary(
    self,
    result: Dict[str, Any]
) -> str:

    data = result["result"]

    if data.get("status") != "success":

        return (
            "There is insufficient data to generate "
            "a reliable business interpretation."
        )

    lines = []

    lines.append(
        "PHARMALENS AI — BUSINESS INSIGHT"
    )

    lines.append("")

    lines.append(
        "WHAT?\n"
        "The analysis was successfully completed "
        "using the relevant PharmaLens intelligence module."
    )

    lines.append("")

    lines.append(
        "KEY DATA"
    )

    for key, value in data.items():

        if key == "status":
            continue

        lines.append(
            f"- {key}: {value}"
        )

    lines.append("")

    lines.append(
        "WHY IT MATTERS\n"
        "The observed performance should be evaluated "
        "against market growth, competition, brand strength, "
        "therapeutic opportunity and commercial execution."
    )

    lines.append("")

    lines.append(
        "WHAT NEXT?\n"
        "Use the relevant PharmaLens predictive and strategic "
        "modules to evaluate forecast, competition, GTM and "
        "recommendation implications."
    )

    return "\n".join(lines)
```

```python
def ask(
    self,
    question: str,
    parameters: Optional[Dict[str, Any]] = None
):

    result = self.route_question(
        question,
        parameters
    )

    return self.business_summary(result)
```
